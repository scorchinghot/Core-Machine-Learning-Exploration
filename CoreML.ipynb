# 🔹 1. Regression (predicting continuous values)

### **Linear Regression**

* **What it is**: A method to model the relationship between input features and a continuous target by fitting a straight line (or hyperplane in higher dimensions).
* **How it works**: Finds the line that minimizes the difference between predicted and actual values (least squares).
* **Strengths**: Simple, interpretable, fast, baseline model.
* **Weaknesses**: Assumes linear relationship, sensitive to outliers, struggles with complex patterns.
* **When to use**: Quick baseline, when relationships are roughly linear and interpretability is key.
* **When not**: Complex nonlinear data, lots of multicollinearity, heavy outliers.

### **Ridge & Lasso Regression**

* **What it is**: Linear regression + regularization to prevent overfitting. Ridge (L2 penalty) shrinks coefficients; Lasso (L1 penalty) can shrink some to zero (feature selection).
* **How it works**: Adds a penalty to large coefficients → forces simpler, more robust models.
* **Strengths**: Handles overfitting, improves generalization, Lasso can auto-select features.
* **Weaknesses**: Still assumes linearity, choice of regularization parameter is crucial.
* **When to use**: High-dimensional data, multicollinearity, need feature selection.
* **When not**: Highly nonlinear patterns.

---

# 🔹 2. Classification (predicting categories)

### **Logistic Regression**

* **What it is**: Linear model for binary (yes/no) classification.
* **How it works**: Uses a sigmoid function to output probabilities instead of raw values.
* **Strengths**: Interpretable, simple, outputs probabilities.
* **Weaknesses**: Linear boundaries only, can underperform on complex data.
* **When to use**: Baseline classifier, when interpretability is important.
* **When not**: Complex non-linear data with many interactions.

### **k-Nearest Neighbors (KNN)**

* **What it is**: Non-parametric classifier that assigns a label based on the majority class of nearest data points.
* **How it works**: Stores all training data, then checks the *k* closest points to classify a new sample.
* **Strengths**: Simple, no training phase, works well with local patterns.
* **Weaknesses**: Slow at prediction, sensitive to irrelevant features and scaling, memory-heavy.
* **When to use**: Small datasets, when decision boundaries are irregular.
* **When not**: Large datasets, high-dimensional data.

### **Decision Tree**

* **What it is**: A tree structure that splits data into groups based on feature thresholds.
* **How it works**: Greedily chooses feature splits that maximize purity (e.g., Gini, entropy).
* **Strengths**: Interpretable, handles non-linearities, no need for scaling.
* **Weaknesses**: Prone to overfitting, unstable (small data changes → big tree changes).
* **When to use**: Tabular data, need interpretability.
* **When not**: Small data with noise, where overfitting is a risk.

### **Random Forest**

* **What it is**: An ensemble of decision trees, combined via averaging/voting.
* **How it works**: Each tree is trained on a random subset of data + features → reduces variance.
* **Strengths**: Robust, handles nonlinearities, less overfitting, feature importance.
* **Weaknesses**: Less interpretable, slower, memory-heavy.
* **When to use**: General-purpose, tabular data, need robustness.
* **When not**: When speed/interpretability is critical.

### **Gradient Boosting (XGBoost/LightGBM)**

* **What it is**: Boosting = sequentially combining weak trees, each correcting the errors of the previous. XGBoost/LightGBM are efficient implementations.
* **How it works**: Each tree is trained to reduce residual error from the previous ensemble.
* **Strengths**: Very powerful, state-of-the-art on many tabular problems, handles complex patterns.
* **Weaknesses**: Many hyperparameters, sensitive to noise, can overfit.
* **When to use**: Kaggle/tabular competitions, structured datasets.
* **When not**: Very small datasets (can overfit), when interpretability is needed.

### **Support Vector Machines (SVM)**

* **What it is**: Finds the hyperplane that maximizes the margin between classes.
* **How it works**: Uses support vectors (boundary cases) + kernels for nonlinear separation.
* **Strengths**: Works well in high-dimensional spaces, robust with clear margins.
* **Weaknesses**: Slow on large datasets, tricky to tune (especially kernels).
* **When to use**: Medium-sized datasets, text classification, bioinformatics.
* **When not**: Very large datasets, noisy overlapping classes.

# 🔹 3. Clustering (unsupervised grouping)

### **K-Means**

* **What it is**: Groups data into *k* clusters by minimizing within-cluster distance.
* **How it works**: Randomly initializes cluster centers, iteratively reassigns points & updates centers.
* **Strengths**: Simple, efficient, widely used.
* **Weaknesses**: Assumes spherical clusters, sensitive to initialization & scaling, must pick *k*.
* **When to use**: Well-separated, roughly round clusters.
* **When not**: Complex shapes, variable density, categorical data.

### **Hierarchical Clustering**

* **What it is**: Builds a hierarchy of clusters (tree-like).
* **How it works**: Agglomerative (merge closest clusters) or divisive (split clusters).
* **Strengths**: No need to predefine *k*, dendrogram visualization.
* **Weaknesses**: Expensive for large datasets, sensitive to noise.
* **When to use**: Small/medium datasets, when you want hierarchy.
* **When not**: Large datasets (scales poorly).

### **DBSCAN (Density-Based Spatial Clustering)**

* **What it is**: Groups dense areas of points, marks sparse points as outliers.
* **How it works**: Expands clusters from high-density regions based on distance + density thresholds.
* **Strengths**: Finds arbitrary-shaped clusters, identifies noise.
* **Weaknesses**: Struggles with varying densities, sensitive to parameters.
* **When to use**: Irregular cluster shapes, anomaly detection.
* **When not**: Clusters with very different densities.

---

# 🔹 4. Dimensionality Reduction / Feature Extraction

### **Principal Component Analysis (PCA)**

* **What it is**: Projects data into fewer dimensions while preserving maximum variance.
* **How it works**: Finds orthogonal directions (principal components) capturing most variance.
* **Strengths**: Reduces dimensionality, removes redundancy, speeds up learning.
* **Weaknesses**: Linear only, loses interpretability, sensitive to scaling.
* **When to use**: Preprocessing, visualization, denoising.
* **When not**: When features have nonlinear relationships.

### **t-SNE / UMAP**

* **What it is**: Nonlinear methods for visualizing high-dimensional data in 2D/3D.
* **How it works**: Preserves local neighborhood structure in a low-dimensional map.
* **Strengths**: Great for visualization, shows clusters clearly.
* **Weaknesses**: Computationally heavy, only for visualization (not feature reduction for models).
* **When to use**: Visual exploration.
* **When not**: As preprocessing for supervised models.

---

# 🔹 5. Ensemble Methods

### **Bagging**

* **What it is**: Train many models on bootstrap samples, average results.
* **How it works**: Reduces variance by combining weak learners.
* **Strengths**: Stabilizes unstable models (like trees), reduces overfitting.
* **Weaknesses**: Less effective if base model is already stable (like linear regression).
* **When to use**: High-variance models, tabular data.
* **When not**: When interpretability is key.

### **Boosting**

* **What it is**: Sequentially improve weak learners by focusing on misclassified points.
* **How it works**: Each learner corrects errors of the previous.
* **Strengths**: Very accurate, handles complex data.
* **Weaknesses**: Sensitive to noise, slower, prone to overfitting.
* **When to use**: Competitive settings, tabular datasets.
* **When not**: Small/noisy datasets.

---

# 🔹 6. Recommendation / Latent Factor Models

### **Matrix Factorization (SVD)**

* **What it is**: Decomposes user-item interaction matrix into latent factors.
* **How it works**: Learns lower-dimensional representation for users & items → predicts missing ratings.
* **Strengths**: Compact, captures hidden preferences.
* **Weaknesses**: Needs lots of data, struggles with new users/items (cold start).
* **When to use**: Recommendation systems with many ratings.
* **When not**: Sparse data, cold-start situations.

### **Collaborative Filtering**

* **What it is**: Recommends items based on similarity between users/items.
* **How it works**: User-based (find similar users) or item-based (find similar items).
* **Strengths**: Simple, intuitive, works without item metadata.
* **Weaknesses**: Cold start, scalability issues.
* **When to use**: Established systems with lots of user-item interactions.
* **When not**: New users/items, sparse datasets.

---

# 🔹 7. Perceptron (Bridge to Deep Learning)

### **Single-Layer Perceptron**

* **What it is**: The simplest neural net — linear classifier with a threshold.
* **How it works**: Inputs → weights → sum → activation → output.
* **Strengths**: Historical foundation, interpretable, simple.
* **Weaknesses**: Only linear boundaries, cannot solve nonlinear problems (like XOR).
* **When to use**: Educational purposes, baseline linear classification.
* **When not**: Nonlinear problems, modern large datasets.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import joblib
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import Ridge, Lasso
```

```python
ratings = pd.read_csv("ml-100k/u.data", sep="\t", names=["user_id", "movie_id", "rating", "timestamp"])
users = pd.read_csv("ml-100k/u.user", sep="|", names=["user_id", "age", "gender", "occupation", "zip_code"])
movies = pd.read_csv("ml-100k/u.item", sep="|", encoding="latin-1", header=None)

movie_cols = [
    "movie_id", "title", "release_date", "video_release_date", "IMDb_URL",
    "unknown", "Action", "Adventure", "Animation", "Children", "Comedy", "Crime",
    "Documentary", "Drama", "Fantasy", "Film-Noir", "Horror", "Musical",
    "Mystery", "Romance", "Sci-Fi", "Thriller", "War", "Western"
]
movies.columns = movie_cols

data = ratings.merge(users, on="user_id").merge(movies, on="movie_id")
```

```python
# Regression target (1–5)
y_reg = data["rating"]

# Classification target (1 if rating >=4, else 0)
y_clf = (data["rating"] >= 4).astype(int)

X = data.drop(columns=["rating", "timestamp", "title", "release_date",
                       "video_release_date", "IMDb_URL", "zip_code"])
```

```python
# Preprocessing
num_features = ["age"]

cat_features = ["gender", "occupation"]

genre_features = [
    "unknown","Action","Adventure","Animation","Children","Comedy","Crime",
    "Documentary","Drama","Fantasy","Film-Noir","Horror","Musical","Mystery",
    "Romance","Sci-Fi","Thriller","War","Western"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_features),
        ("cat", OneHotEncoder(drop="first"), cat_features),
        ("genres", "passthrough", genre_features)
    ]
)

# Apply preprocessing
X_processed = preprocessor.fit_transform(X)
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_processed, y_reg, test_size=0.2, random_state=42)
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_processed, y_clf, test_size=0.2, random_state=42)

print("Feature matrix shape:", X_processed.shape)
print("Regression target shape:", y_reg.shape)
print("Classification target shape:", y_clf.shape)

joblib.dump((X_train_reg, X_test_reg, y_train_reg, y_test_reg), "regression_data.pkl")
joblib.dump((X_train_clf, X_test_clf, y_train_clf, y_test_clf), "classification_data.pkl")
joblib.dump(preprocessor, "preprocessor.pkl")
```

```python
# Linear Regression
X_train_reg, X_test_reg, y_train_reg, y_test_reg = joblib.load("regression_data.pkl")

lin_reg = LinearRegression()
lin_reg.fit(X_train_reg, y_train_reg)

y_pred = lin_reg.predict(X_test_reg)

mse = mean_squared_error(y_test_reg, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_reg, y_pred)

print("Linear Regression Results:")
print(f"  RMSE: {rmse:.4f}")
print(f"  R² Score: {r2:.4f}")
```

```python
# Inspect Coefficients
preprocessor = joblib.load("preprocessor.pkl")
feature_names = []

feature_names += ["age"]

# categorical (from OneHotEncoder)
cat_encoder = preprocessor.named_transformers_["cat"]
cat_feature_names = cat_encoder.get_feature_names_out(["gender", "occupation"])
feature_names += list(cat_feature_names)

feature_names += [
    "unknown","Action","Adventure","Animation","Children","Comedy","Crime",
    "Documentary","Drama","Fantasy","Film-Noir","Horror","Musical","Mystery",
    "Romance","Sci-Fi","Thriller","War","Western"
]

coef_df = pd.DataFrame({
    "Genres": feature_names,
    "Coefficient": lin_reg.coef_
}).sort_values(by="Coefficient", ascending=False)

print("\nTop 10 genres by weight:")
print(coef_df.head(10))
```

```python
# Ridge Regression
ridge = Ridge(alpha=1.0)
ridge.fit(X_train_reg, y_train_reg)
ridge_pred = ridge.predict(X_test_reg)

ridge_rmse = np.sqrt(mean_squared_error(y_test_reg, ridge_pred))
ridge_r2 = r2_score(y_test_reg, ridge_pred)

print("Ridge Regression Results:")
print(f"  RMSE: {ridge_rmse:.4f}")
print(f"  R² Score: {ridge_r2:.4f}")

# Lasso Regression
lasso = Lasso(alpha=0.001)
lasso.fit(X_train_reg, y_train_reg)
lasso_pred = lasso.predict(X_test_reg)

lasso_rmse = np.sqrt(mean_squared_error(y_test_reg, lasso_pred))
lasso_r2 = r2_score(y_test_reg, lasso_pred)

print("\nLasso Regression Results:")
print(f"  RMSE: {lasso_rmse:.4f}")
print(f"  R² Score: {lasso_r2:.4f}")

# Lasso genre selection
coef_df_lasso = pd.DataFrame({
    "Genres": feature_names,
    "Coefficient": lasso.coef_
}).sort_values(by="Coefficient", ascending=False)

print("\nTop 10 Genres kept by Lasso:")
print(coef_df_lasso[coef_df_lasso['Coefficient'] != 0].head(10))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

X_train_clf, X_test_clf, y_train_clf, y_test_clf = joblib.load("classification_data.pkl")

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_clf, y_train_clf)

y_pred = log_reg.predict(X_test_clf)

acc = accuracy_score(y_test_clf, y_pred)
print("Logistic Regression Accuracy:", acc)

print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Key parameters to know:
# n_neighbors: (default=5) Number of neighbors to look at. Too small = noisy, too large = oversmooth.
# weights: 'uniform' (all neighbors equal) or 'distance' (closer neighbors have more influence).
# metric: distance measure, usually 'minkowski' (which is Euclidean when p=2).
# p: power parameter for Minkowski distance. p=1 -> Manhattan, p=2 -> Euclidean.
# algorithm: how neighbors are searched ('auto', 'ball_tree', 'kd_tree', 'brute').

knn = KNeighborsClassifier(
    n_neighbors=5,
    weights='distance', # Make closer neighbors more important
    metric='minkowski',
    p=2                 # 2 = Euclidean distance
)

knn.fit(X_train_clf, y_train_clf)

y_pred_knn = knn.predict(X_test_clf)

acc_knn = accuracy_score(y_test_clf, y_pred_knn)
print("KNN Accuracy:", acc_knn)

print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_knn))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_knn))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Important parameters:
# criterion: 'gini' (default) or 'entropy' (information gain) — how to measure purity.
# max_depth: limit the depth of the tree (prevents overfitting).
# min_samples_split: minimum samples required to split a node.
# min_samples_leaf: minimum samples required at a leaf.
# max_features: number of features to consider when splitting.

tree_clf = DecisionTreeClassifier(
    criterion='gini',   # Try 'entropy' too
    max_depth=10,       # Limit depth to avoid overfitting
    min_samples_split=20, # Don't split tiny nodes
    random_state=42
)

tree_clf.fit(X_train_clf, y_train_clf)

y_pred_tree = tree_clf.predict(X_test_clf)

print("Decision Tree Accuracy:", accuracy_score(y_test_clf, y_pred_tree))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_tree))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_tree))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Important parameters:
# n_estimators: number of trees in the forest (more = better, but slower).
# max_depth: maximum depth of each tree (control overfitting).
# min_samples_split: minimum samples to split a node.
# min_samples_leaf: minimum samples per leaf.
# max_features: number of features to consider per split (default: 'sqrt' for classification).
# bootstrap: whether to use bootstrapped samples (default=True).

rf_clf = RandomForestClassifier(
    n_estimators=100,     # 100 trees is a good starting point
    max_depth=None,       # Let trees grow fully (can tune later)
    min_samples_split=10, # Prevent splitting on tiny subsets
    min_samples_leaf=5,   # Each leaf must have at least 5 samples
    max_features='sqrt',  # Random subset of features for each split
    bootstrap=True,       # Bootstrap sampling (bagging)
    random_state=42,
    n_jobs=-1             # Use all CPU cores
)

rf_clf.fit(X_train_clf, y_train_clf)

y_pred_rf = rf_clf.predict(X_test_clf)


print("Random Forest Accuracy:", accuracy_score(y_test_clf, y_pred_rf))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_rf))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_rf))

# One of the best things about Random Forest: Feature Importances.
# It tells you which features (genres, occupations, etc.) the forest relied on most.
feature_importances = rf_clf.feature_importances_

feat_imp_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": feature_importances
}).sort_values(by="Importance", ascending=False)

print("\nTop 10 Most Important Features:")
print(feat_imp_df.head(10))
```

```python
from sklearn.ensemble import GradientBoostingClassifier

# n_estimators: number of boosting stages (more = stronger but slower)
# learning_rate: shrinkage factor (0.1 default, lower = slower learning)
# max_depth: max depth of each tree (weak learners are usually shallow)
# min_samples_split / min_samples_leaf: control overfitting
# subsample: fraction of samples per tree (stochastic gradient boosting)
# max_features: number of features considered at each split

gb_clf = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=3,        # shallow trees (weak learners)
    subsample=0.8,      # stochastic gradient boosting
    max_features='sqrt',# subset of features
    random_state=42
)

gb_clf.fit(X_train_clf, y_train_clf)

y_pred_gb = gb_clf.predict(X_test_clf)

print("Gradient Boosting (sklearn) Accuracy:", accuracy_score(y_test_clf, y_pred_gb))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_gb))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_gb))

feat_imp_gb = pd.DataFrame({
    "Feature": feature_names,
    "Importance": gb_clf.feature_importances_
}).sort_values(by="Importance", ascending=False)

print("\nTop 10 Features (Gradient Boosting sklearn):")
print(feat_imp_gb.head(10))
```

```python
import xgboost as xgb
import lightgbm as lgb

# Key Hyperparameters (general)
# n_estimators: number of trees (more = stronger, but slower).
# learning_rate: shrinkage factor (lower = slower learning, but more accurate).
# max_depth: depth of each tree.
# subsample: fraction of data for each tree (prevents overfitting).
# colsample_bytree (XGBoost/LightGBM): fraction of features per tree.
# min_child_weight / min_child_samples: minimum samples per leaf.

xgb_clf = xgb.XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

xgb_clf.fit(X_train_clf, y_train_clf)
y_pred_xgb = xgb_clf.predict(X_test_clf)

print("XGBoost Accuracy:", accuracy_score(y_test_clf, y_pred_xgb))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_xgb))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_xgb))

lgb_clf = lgb.LGBMClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

lgb_clf.fit(X_train_clf, y_train_clf)
y_pred_lgb = lgb_clf.predict(X_test_clf)

print("\nLightGBM Accuracy:", accuracy_score(y_test_clf, y_pred_lgb))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_lgb))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_lgb))

feat_imp_xgb = pd.DataFrame({
    "Feature": feature_names,
    "Importance": xgb_clf.feature_importances_
}).sort_values(by="Importance", ascending=False)

print("Top 10 Features XGBoost:")
print(feat_imp_xgb.head(10))
```

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.preprocessing import StandardScaler

# Optional: scale features for clustering (important for distance-based algorithms)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_processed)

# K-Means Clustering
# Key parameters:
# n_clusters: number of clusters to find.
# init: method to initialize centroids ('k-means++' recommended, 'random' possible).
# n_init: number of initializations to run (higher = better chance to converge to global optimum).
# max_iter: maximum iterations per run.
# random_state: for reproducibility.

kmeans = KMeans(
    n_clusters=5,       # choose number of clusters
    init='k-means++',   # smart centroid initialization
    n_init=10,          # run algorithm 10 times with different seeds
    max_iter=300,       # max iterations per run
    random_state=42
)
kmeans_labels = kmeans.fit_predict(X_scaled)
print("KMeans Cluster Counts:", np.bincount(kmeans_labels))

# Hierarchical (Agglomerative) Clustering
# Key parameters:
# n_clusters: number of clusters to find (optional if using distance_threshold)
# metric: distance metric ('euclidean', 'manhattan', etc.) - Corrected from affinity
# linkage: 'ward' (minimize variance), 'complete', 'average', 'single'
# distance_threshold: set instead of n_clusters to cut dendrogram at this height

hier_clust = AgglomerativeClustering(
    n_clusters=5,      # number of clusters to form
    metric='euclidean',  # distance metric
    linkage='ward'         # method to merge clusters
)
hier_labels = hier_clust.fit_predict(X_scaled)
print("Hierarchical Cluster Counts:", np.bincount(hier_labels))

# DBSCAN (Density-Based Spatial Clustering)
# Key parameters:
# eps: maximum distance between two samples to be considered neighbors
# min_samples: minimum number of points to form a dense region
# metric: distance metric ('euclidean' by default)
# DBSCAN does NOT require number of clusters in advance.

dbscan = DBSCAN(
    eps=3,             # distance threshold for neighborhood
    min_samples=10,    # minimum points to form a dense cluster
    metric='euclidean'
)
dbscan_labels = dbscan.fit_predict(X_scaled)
unique_labels = set(dbscan_labels)
print("DBSCAN Cluster Counts:", {label: sum(dbscan_labels == label) for label in unique_labels})
```

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap
import matplotlib.pyplot as plt

# Optional: scale features before dimensionality reduction
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(X_processed)

# PCA (Principal Component Analysis)
# Key parameters:
# n_components: number of dimensions to reduce to
# whiten: decorrelates components and scales variance to 1 (optional)
# svd_solver: 'auto', 'full', 'arpack', 'randomized' (affects speed and accuracy)

pca = PCA(
    n_components=2,   # reduce to 2D for visualization
    whiten=False,     # keep original variance scaling
    svd_solver='auto'
)
X_pca = pca.fit_transform(X_scaled)
print("PCA explained variance ratio:", pca.explained_variance_ratio_)

plt.figure(figsize=(6,5))
plt.scatter(X_pca[:,0], X_pca[:,1], alpha=0.5)
plt.title("PCA 2D Projection")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()

# t-SNE (t-distributed Stochastic Neighbor Embedding)
# Key parameters:
# n_components: dimensions to reduce to (usually 2 or 3 for visualization)
# perplexity: balances attention between local and global structure (typical 5-50)
# learning_rate: controls speed of optimization (default 200)
# n_iter: number of iterations (more = better convergence)
# random_state: for reproducibility

tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate=200,
    n_iter=1000,
    random_state=42
)
X_tsne = tsne.fit_transform(X_scaled)

plt.figure(figsize=(6,5))
plt.scatter(X_tsne[:,0], X_tsne[:,1], alpha=0.5)
plt.title("t-SNE 2D Projection")
plt.show()

# UMAP (Uniform Manifold Approximation and Projection)
# Key parameters:
# n_neighbors: controls local neighborhood size (small = local, large = global structure)
# n_components: dimensions to reduce to
# min_dist: controls how tightly points are packed together
# metric: distance metric ('euclidean' by default)

umap_reducer = umap.UMAP(
    n_neighbors=15,
    n_components=2,
    min_dist=0.1,
    metric='euclidean',
    random_state=42
)
X_umap = umap_reducer.fit_transform(X_scaled)

plt.figure(figsize=(6,5))
plt.scatter(X_umap[:,0], X_umap[:,1], alpha=0.5)
plt.title("UMAP 2D Projection")
plt.show()
```

```python
from sklearn.ensemble import BaggingClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier

#  Bagging (Bootstrap Aggregation)
# Idea: Train many base learners (usually trees) on **bootstrapped subsets** of the data
# and average predictions (for regression) or vote (for classification).
# Reduces variance, prevents overfitting.

# Key parameters:
# estimator: the model to train on each bootstrap sample (default=DecisionTreeClassifier) - Corrected from base_estimator
# n_estimators: number of base learners
# max_samples: fraction of samples to use per base learner
# max_features: fraction of features to use per base learner
# bootstrap: whether to sample with replacement
# n_jobs: number of parallel jobs

bag_clf = BaggingClassifier(
    estimator=DecisionTreeClassifier(max_depth=5), # Corrected parameter name
    n_estimators=50,
    max_samples=0.8,
    max_features=1.0,
    bootstrap=True,
    n_jobs=-1,
    random_state=42
)

bag_clf.fit(X_train_clf, y_train_clf)

y_pred_bag = bag_clf.predict(X_test_clf)

print("Bagging Classifier Accuracy:", accuracy_score(y_test_clf, y_pred_bag))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_bag))

#  Boosting (AdaBoost / Gradient Boosting)
# Idea: Sequentially train weak learners. Each new learner focuses on the **errors of previous ones**.
# Reduces bias, can overfit if not regularized.

# AdaBoost
# Key parameters:
# base_estimator: model to boost (default=DecisionTreeClassifier(max_depth=1))
# n_estimators: number of weak learners
# learning_rate: contribution of each learner
# algorithm: 'SAMME' or 'SAMME.R' (real-valued boosting)

ada_clf = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=1), # base_estimator is still used in AdaBoostClassifier
    n_estimators=100,
    learning_rate=0.5,
    algorithm='SAMME', # Changed from 'SAMME.R' to 'SAMME'
    random_state=42
)

ada_clf.fit(X_train_clf, y_train_clf)
y_pred_ada = ada_clf.predict(X_test_clf)

print("\nAdaBoost Accuracy:", accuracy_score(y_test_clf, y_pred_ada))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_ada))

# Already covered previously, but key takeaway:
# - Sequential learning
# - Shallow trees as weak learners
# - Learning rate and n_estimators control bias-variance tradeoff
gb_clf = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=3,
    subsample=0.8,
    max_features='sqrt',
    random_state=42
)
gb_clf.fit(X_train_clf, y_train_clf)
y_pred_gb = gb_clf.predict(X_test_clf)

print("\nGradient Boosting Accuracy:", accuracy_score(y_test_clf, y_pred_gb))
print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_gb))
```
